```python
import pandas as pd

pairs_to_check = [
    ("your_persona_original", "partner_persona_original"),
    ("your_persona_original", "your_persona_revised"),
    ("partner_persona_original", "partner_persona_revised"),
    ("your_persona_revised", "partner_persona_revised"),
]

def to_hashable(value):
    if isinstance(value, dict):
        return tuple(sorted((key, to_hashable(item)) for key, item in value.items()))
    if isinstance(value, (list, tuple)):
        return tuple(to_hashable(item) for item in value)
    return value

df = dataset.to_pandas()
results = []

for left_column, right_column in pairs_to_check:
    values = zip(df[left_column], df[right_column])
    pairs = [(to_hashable(left), to_hashable(right)) for left, right in values]
    unique_count = len(set(pairs))
    results.append(
        {
            "par": f"({left_column}, {right_column})",
            "total": len(pairs),
            "unicos": unique_count,
            "duplicatas": len(pairs) - unique_count,
            "e_unico": unique_count == len(pairs),
        }
    )

pd.DataFrame(results)
```

```python
mappings_to_check = [
    ("your_persona_original", "your_persona_revised"),
    ("partner_persona_original", "partner_persona_revised"),
]

mapping_results = []
conflicting_rows = []

for original_column, revised_column in mappings_to_check:
    revision_counts = df.groupby(original_column, dropna=False)[revised_column].nunique(dropna=False)
    originals_with_multiple_revisions = revision_counts[revision_counts > 1]

    mapping_results.append(
        {
            "mapeamento": f"{original_column} -> {revised_column}",
            "originais_unicos": len(revision_counts),
            "originais_com_multiplas_revisoes": len(originals_with_multiple_revisions),
            "max_revisoes_por_original": revision_counts.max(),
            "e_funcao": (revision_counts <= 1).all(),
        }
    )

    if not originals_with_multiple_revisions.empty:
        conflicts = df[df[original_column].isin(originals_with_multiple_revisions.index)][
            [original_column, revised_column]
        ].drop_duplicates()
        conflicts.insert(0, "mapeamento", f"{original_column} -> {revised_column}")
        conflicting_rows.append(conflicts)

mapping_summary = pd.DataFrame(mapping_results)
display(mapping_summary)

if conflicting_rows:
    display(pd.concat(conflicting_rows, ignore_index=True))
else:
    print("Nenhuma persona original possui mais de uma versão revisada.")
```

```python
persona_columns = [column for column in df.columns if "persona" in column.lower()]

def is_missing(value):
    if value is None:
        return True
    if isinstance(value, str):
        normalized_value = value.strip()
        return not normalized_value or normalized_value == "[]"
    if isinstance(value, (list, tuple, dict, set)):
        return len(value) == 0
    try:
        return bool(pd.isna(value))
    except (TypeError, ValueError):
        return False

missing_mask = pd.DataFrame(
    {column: df[column].map(is_missing) for column in persona_columns},
    index=df.index,
)

validation_summary = pd.DataFrame(
    {
        "preenchidos": (~missing_mask).sum(),
        "ausentes_ou_vazios": missing_mask.sum(),
        "todos_preenchidos": ~missing_mask.any(),
    }
)

invalid_rows = pd.DataFrame(
    {
        "colunas_ausentes_ou_vazias": missing_mask.apply(
            lambda row: [column for column, missing in row.items() if missing],
            axis=1,
        ),
        "valores_invalidos": missing_mask.apply(
            lambda row: {
                column: df.at[row.name, column]
                for column, missing in row.items()
                if missing
            },
            axis=1,
        ),
    }
).loc[missing_mask.any(axis=1)]

dataset_is_valid = invalid_rows.empty
print(f"Dataset válido: {dataset_is_valid}")
print(f"Registros incompletos: {len(invalid_rows)} de {len(df)}")
display(validation_summary)

if not dataset_is_valid:
    display(invalid_rows)
```
